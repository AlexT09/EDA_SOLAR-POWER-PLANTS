## 1.3 Diccionario de las predictoras

| Variable | Tipo | Descripción |
|---|---|---|
| `longitude`, `latitude` | Numérica (°) | Coordenadas geográficas de la planta |

| `wind_speed` | Numérica (m/s) | Velocidad media del viento |
| `wind_direction` | Numérica (°) | Dirección predominante del viento |
| `slope_type` | Categórica (6) | Clase de pendiente |
| `curvature_type` | Categórica (3) | Clase de curvatura |
| `aspect_type` | Categórica (9) | Orientación cardinal |
